```python
import numpy as np
from collections import defaultdict

all_metrics      = ["rmse", "mRMSE", "hit_rate", "csi", "f2_score", "f3_score",
                    "flops", "parameters", "peak_gpu_memory", "event_time", "timestep_time"]
temporal_metrics = ["rmse", "mRMSE", "hit_rate", "csi", "f2_score"]
eventl_level_metrics    = ["event_time", "peak_volume_timing_error"]
model_metrics    = ["flops", "parameters", "peak_gpu_memory"]

metrics_df  = pd.read_csv(model_metrics_file)
#select the last 5 rows of the dataframe
metrics_df = metrics_df.tail(5)
summary_metrics = {}

# --step 1: Summary Metrics Table--
for metric in temporal_metrics:
    all_runs = metrics_df[metric]
    event_metrics = {}
    for idx, metric_values in enumerate(all_runs):
        #convert the string representation of the dictonary back to an actual dictionary
        metric_values_dict = eval(metric_values)
        #each key represents an event, and the value is a list of metric values for that event across all timesteps
        for event, timestep_values in metric_values_dict.items():
            #convert the timestep values to a numpy array for easier manipulation
            timestep_values_array = np.array(timestep_values)
            #average the metric values across all timesteps for this event
            average_metric_value = np.mean(timestep_values_array)
            #store the average metric value in the summary metrics dictionary
            if event not in event_metrics:
                event_metrics[event] = []
            event_metrics[event].append(average_metric_value)
    for event, metric_values in event_metrics.items():
        mean_avg_metric_value = np.mean(metric_values)
        std_avg_metric_value = np.std(metric_values)
        if event not in summary_metrics:
            summary_metrics[event] = {}
        summary_metrics[event][metric] = {"mean": mean_avg_metric_value, "std": std_avg_metric_value}
        
for metric in eventl_level_metrics:
    all_runs = metrics_df[metric]
    metrics = {}
    for idx, metric_values in enumerate(all_runs):
        #convert the string representation of the dictonary back to an actual dictionary
        #each key represents an event, and the value is a single metric value for that event
        metric_values_dict = eval(metric_values)  
        for event, metric_value in metric_values_dict.items():
            #store the metric value in the summary metrics dictionary
            if event not in metrics:
                metrics[event] = []
            metrics[event].append(metric_value)

    for event, metric_values in metrics.items():
        mean_metric_value = np.mean(metric_values)
        std = np.std(metric_values)
        summary_metrics[event][metric] = {"mean": mean_metric_value, "std": std}  # No std for event-level metrics

summary_metrics
#create a dataframe from the summary metrics dictionary
#use the key as the first column, and the metric mean and std as the second and third columns
summary_metrics_df = pd.DataFrame.from_dict(summary_metrics, orient='index')
summary_metrics_df.reset_index(inplace=True)
summary_metrics_df.rename(columns={"index": "event"}, inplace=True)
summary_metrics_df.head(5)
```

```python
from collections import defaultdict
#Step 2: Carilse Temporal Evoluation of metrics
temporal_metrics = ["rmse", "mRMSE", "hit_rate", "csi", "f2_score"]
metrics_df  = pd.read_csv(model_metrics_file)
metrics_df = metrics_df.tail(5)

# First need to organise the data
collected = defaultdict(lambda: defaultdict(list))  # {metric: {event: [values across runs]}}

for idx, row in metrics_df.iterrows():
    data = row.to_dict()  # Convert the row to a dictionary for easier access

    # Temporal: {event: [v, v, ...]}
    for metric in temporal_metrics:
        if metric in data:
            metric_data = data[metric]
            if isinstance(metric_data, str):
                metric_data = eval(metric_data)  # Convert string representation of dict to actual dict
            for event, values in metric_data.items():
                collected[metric][event].append(np.array(values, dtype=float))

averaged_metrics = {} #key event -> value dict of metric -> mean and std   
for metric, events in collected.items():
    for event, values in events.items():
        # Convert list of arrays to a 2D array (runs x timesteps)
        values_array = np.array(values)
        mean_values = np.mean(values_array, axis=0)  # Average across runs
        std_values = np.std(values_array, axis=0)    # Std across runs
        if event not in averaged_metrics:
            averaged_metrics[event] = {}
        averaged_metrics[event][metric] = {"mean": mean_values, "std": std_values}


#convert to a dataframe and save it
averaged_metrics_df = pd.DataFrame.from_dict(averaged_metrics, orient='index')
averaged_metrics_df.reset_index(inplace=True)
averaged_metrics_df.rename(columns={"index": "event"}, inplace=True)
averaged_metrics_df.head(5)

#save the dataframe to a csv file
averaged_metrics_df.to_csv(os.path.join(data_dir, model_name, "averaged_temporal_metrics.csv"), index=False)
```

```python
import numpy as np
from collections import defaultdict

all_metrics      = ["rmse", "mRMSE", "hit_rate", "csi", "f2_score", "f3_score",
                    "flops", "parameters", "peak_gpu_memory", "event_times", "timestep_times"]
temporal_metrics = ["rmse", "mRMSE", "hit_rate", "csi", "f2_score", "f3_score", "timestep_times"]
event_metrics    = ["event_times", "peak_gpu_memory"]
model_metrics    = ["flops", "parameters"]

# ── Step 1: collect raw values from all runs ──────────────────────────────────
# Structure: collected[metric][event] = list of arrays/values, one per run

collected = defaultdict(lambda: defaultdict(list))

for idx, row in metrics_df.iterrows():
    data = row.to_dict()  # Convert the row to a dictionary for easier access

    # Temporal: {event: [v, v, ...]}
    for metric in temporal_metrics:
        if metric in data:
            metric_data = data[metric]
            if isinstance(metric_data, str):
                metric_data = eval(metric_data)  # Convert string representation of dict to actual dict
            for event, values in metric_data.items():
                collected[metric][event].append(np.array(values, dtype=float))

    # Event-level scalars: {event: scalar}  OR  scalar
    for metric in event_metrics:
        if metric in data:
            val = data[metric]
            if isinstance(val, str):
                val = eval(val)  # Convert string representation of dict to actual dict
            if isinstance(val, dict):
                for event, v in val.items():
                    collected[metric][event].append(float(v))
            else:
                collected[metric]["all"].append(float(val))

    # Model-level scalars: single value
    for metric in model_metrics:
        if metric in data:
            collected[metric]["all"].append(float(data[metric]))


# ── Step 2: compute mean and std ──────────────────────────────────────────────

aggregated = defaultdict(dict)

for metric, events in collected.items():
    for event, values in events.items():

        if isinstance(values[0], np.ndarray):
            # Temporal — align lengths then stack
            min_len = min(len(v) for v in values)
            mat = np.stack([v[:min_len] for v in values])  # (n_runs, T)
            aggregated[metric][event] = {
                "mean": np.mean(mat, axis=0),  # (T,)
                "std":  np.std(mat,  axis=0),
                "all":  mat,
            }
        else:
            # Scalar
            arr = np.array(values)
            aggregated[metric][event] = {
                "mean": np.mean(arr),
                "std":  np.std(arr),
                "all":  arr,
            }


# ── Step 3: quick sanity check ────────────────────────────────────────────────

for metric in all_metrics:
    if metric in aggregated:
        for event, stats in aggregated[metric].items():
            shape = stats["mean"].shape if hasattr(stats["mean"], "shape") else "(scalar)"
            print(f"  {metric:<20} {event:<12} mean shape: {shape}")
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
import pandas as pd
import os

GRAPH_OUTPUT_DIR = f"/home/91/23016891/projects/carlisle/out/carlisle"

# ── Stage classification ──────────────────────────────────────────────────────
stage_labels = {0: "Early", 1: "Mid", 2: "Late"}
stage_colors = {
    "Early": "#000000",
    "Mid":   "#1900ff",
    "Late":  "#ff0000",
}

# ── Build DataFrames ──────────────────────────────────────────────────────────
def build_df(data, metric_name):
    records = []
    for event in data.keys():
        values = data[event]["mean"]  # shape (T,)
        T = len(values)
        for t, val in enumerate(values):
            norm_t = t / T
            stage  = min(int(norm_t * 3), 2)
            records.append({
                "event":    event,
                "timestep": t,
                "norm_t":   norm_t,
                "stage":    stage_labels[stage],
                metric_name: val,
            })
    return pd.DataFrame(records)

rmse_df  = build_df(rmse_data,  "rmse")
mrmse_df = build_df(mrmse_data, "mrmse")

# ── Colour mappings ───────────────────────────────────────────────────────────
events       = rmse_df["event"].unique()
cmap_events  = plt.cm.tab20
event_colors = {e: cmap_events(i / len(events)) for i, e in enumerate(events)}

# ── Box plot helper ───────────────────────────────────────────────────────────
def draw_boxplot(ax, values, ylabel):
    ax.boxplot(
        values, positions=[1], vert=True,
        patch_artist=True, widths=0.35,
        boxprops=dict(facecolor="white", edgecolor="#2166ac", linewidth=1.2),
        medianprops=dict(color="#d6604d", linewidth=2),
        whiskerprops=dict(color="#2166ac", linewidth=1.0),
        capprops=dict(color="#2166ac", linewidth=1.0),
        flierprops=dict(visible=False),
    )
    ax.set_xticks([])
    ax.set_xlim(0.7, 1.5)
    ax.set_ylabel(ylabel, fontsize=12)
    ax.yaxis.grid(True, linestyle="--", linewidth=0.5, color="#cccccc", alpha=0.8)
    ax.set_axisbelow(True)
    for spine in ["top", "right", "bottom"]:
        ax.spines[spine].set_visible(False)
    ax.spines["left"].set_color("#2c3e50")
    ax.tick_params(colors="#2c3e50", labelsize=10)

# ── Figure: 2 rows x 2 columns ───────────────────────────────────────────────
fig, axes = plt.subplots(2, 2, figsize=(10, 10), sharey="row")
fig.suptitle(f"{study_name} — {model_name}", fontsize=12, y=1.01)

stage_handles = [
    mpatches.Patch(color=stage_colors[s], label=s)
    for s in ["Early", "Mid", "Late"]
]

for row, (df, metric, ylabel) in enumerate([
    (rmse_df,  "rmse",  "RMSE (m)"),
    (mrmse_df, "mrmse", "mRMSE (m)"),
]):
    # ── Panel (a): coloured by event ─────────────────────────────────────────
    draw_boxplot(axes[row, 0], df[metric].values, ylabel)
    np.random.seed(0)
    for event, grp in df.groupby("event"):
        x_jitter = np.random.normal(1, 0.04, size=len(grp))
        axes[row, 0].scatter(x_jitter, grp[metric].values,
                             color=event_colors[event], s=8, alpha=0.5,
                             edgecolors="none", zorder=2, label=str(event))

    axes[row, 0].legend(fontsize=7, frameon=True, loc="upper right",
                        markerscale=2, title="Event", title_fontsize=8)
    axes[row, 0].set_title(f"({'ab'[row]}) {metric.upper()} — coloured by event",
                           loc="left", fontsize=11, pad=6)

    # ── Panel (b): coloured by flood stage ───────────────────────────────────
    draw_boxplot(axes[row, 1], df[metric].values, "")
    np.random.seed(0)
    for stage, grp in df.groupby("stage"):
        x_jitter = np.random.normal(1, 0.04, size=len(grp))
        axes[row, 1].scatter(x_jitter, grp[metric].values,
                             color=stage_colors[stage], s=8, alpha=0.5,
                             edgecolors="none", zorder=2)

    axes[row, 1].legend(handles=stage_handles, fontsize=8, frameon=True,
                        loc="upper right", title="Flood stage", title_fontsize=8)
    axes[row, 1].set_title(f"({'cd'[row]}) {metric.upper()} — coloured by flood stage",
                           loc="left", fontsize=11, pad=6)

plt.tight_layout()
plt.savefig(os.path.join(GRAPH_OUTPUT_DIR,
            f"fig_rmse_mrmse_boxplot_{study_name}_{model_name}.png"),
            dpi=300, bbox_inches="tight")
plt.show()
```
